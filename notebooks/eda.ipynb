# Exploratory Data Analysis (EDA) for EEG Authentication

This notebook is dedicated to performing exploratory data analysis on the EEG dataset. We will visualize the data, summarize statistics, and understand the distribution of features.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load the processed EEG data
eeg_data = np.load('../data/processed/processed_eeg.npy')

# Convert to DataFrame for easier analysis
eeg_df = pd.DataFrame(eeg_data)
eeg_df.head()
```

```python
# Summary statistics
eeg_summary = eeg_df.describe()
eeg_summary
```

```python
# Visualize the distribution of the first channel
plt.figure(figsize=(10, 6))
sns.histplot(eeg_df[0], bins=30, kde=True)
plt.title('Distribution of EEG Channel 1')
plt.xlabel('Amplitude')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plotting the first 5 seconds of all channels
sampling_rate = 160  # Hz
time = np.linspace(0, 5, 5 * sampling_rate)

plt.figure(figsize=(10, 6))
for i in range(eeg_df.shape[1]):
    plt.plot(time, eeg_df.iloc[:len(time), i] + i * 200, label=f'Channel {i + 1}')

plt.title('EEG Signals (First 5 Seconds)')
plt.xlabel('Time (seconds)')
plt.ylabel('Amplitude (shifted)')
plt.legend()
plt.show()
```
