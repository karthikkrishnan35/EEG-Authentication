# Model Evaluation

This notebook is dedicated to evaluating the performance of the trained EEG binary classification model. We will use various metrics to assess the model's effectiveness.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report
from src.models.train import load_model
from src.evaluation.metrics import calculate_metrics

# Load the test data
test_data = pd.read_csv('data/processed/test_data.csv')
X_test = test_data.drop('label', axis=1)
y_test = test_data['label']

# Load the trained model
model = load_model('models/trained_model.h5')

# Make predictions
y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)

# Calculate metrics
conf_matrix = confusion_matrix(y_test, y_pred_classes)
report = classification_report(y_test, y_pred_classes)

# Display results
plt.figure(figsize=(8, 6))
plt.imshow(conf_matrix, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.colorbar()
tick_marks = np.arange(len(np.unique(y_test)))
plt.xticks(tick_marks, np.unique(y_test))
plt.yticks(tick_marks, np.unique(y_test))
plt.ylabel('True label')
plt.xlabel('Predicted label')
plt.show()

# Print classification report
print(report)
```
